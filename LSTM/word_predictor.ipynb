```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Dropout, Embedding
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences 
from tensorflow.keras.utils import to_categorical

tf.random.set_seed(34)
```

```python
with open("dataset2.txt", "r") as f:
    dataset = f.read()
```

```python
tokenizer = Tokenizer() 
tokenizer.fit_on_texts([dataset])
TOTAL_WORDS = len(tokenizer.word_index)+1
```

```python
# Create the dataset for LSTM
tokenized_sentences_array = []
MAX_LEN = 0

for sentences in dataset.split("\n"):
    tokenized_sentence = tokenizer.texts_to_sequences([sentences])[0]
    MAX_LEN = max(MAX_LEN, len(tokenized_sentence))
    
    if tokenized_sentence == []: continue

    for token in range(1, len(tokenized_sentence)):
        n_gram = tokenized_sentence[:token+1]
        tokenized_sentences_array.append(n_gram) 

tokenized_sentences_array
```

```python
padded_sentences_array = pad_sequences(tokenized_sentences_array, maxlen = MAX_LEN, padding = "pre") 
```

```python
X = padded_sentences_array[:,:-1]
y = padded_sentences_array[:,-1]
print(X.shape)
print(y.shape)
```

```python
y = to_categorical(y, num_classes=TOTAL_WORDS)
y.shape
```

```python
model = Sequential(
    [
        Embedding(TOTAL_WORDS, 128, input_length=MAX_LEN),
        LSTM(128, return_sequences=True),
        Dropout(0.2),
        LSTM(64), 
        Dropout(0.2),
        Dense(TOTAL_WORDS, activation='softmax')
    ]
)
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()
```

```python
result = model.fit(X,y,epochs=5)
```
